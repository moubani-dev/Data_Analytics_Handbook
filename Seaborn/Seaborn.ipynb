```python
# Program 1: Import required libraries and load a sample dataset

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load Seaborn's built-in tips dataset
df = sns.load_dataset("tips")

# Show first 5 rows
print(df.head())
```

```python
# Program 2: Display basic information about the dataset

import seaborn as sns

df = sns.load_dataset("tips")

# Display number of rows and columns
print("Shape of dataset:", df.shape)

# Display column names
print("Columns:", df.columns)

# Display data types
print(df.dtypes)
```

```python
# Program 3: Create a simple line plot

import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

# Line plot between total_bill and tip
sns.lineplot(x="total_bill", y="tip", data=df)

plt.title("Line Plot of Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
# Program 4: Create a bar plot

import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

# Bar plot showing average total bill by day
sns.barplot(x="day", y="total_bill", data=df)

plt.title("Average Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Average Total Bill")
plt.show()
```

```python
# Program 5: Create a count plot

import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

# Count how many records are present for each day
sns.countplot(x="day", data=df)

plt.title("Count of Records by Day")
plt.xlabel("Day")
plt.ylabel("Count")
plt.show()
```

```python
# Program 6: Create a scatter plot

import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

# Scatter plot between total bill and tip
sns.scatterplot(x="total_bill", y="tip", data=df)

plt.title("Scatter Plot of Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
# Program 7: Create a histogram

import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

# Histogram of total_bill
sns.histplot(df["total_bill"], bins=10, kde=False)

plt.title("Histogram of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
# Program 8: Histogram with KDE curve

import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

# Histogram with KDE
sns.histplot(df["total_bill"], bins=10, kde=True)

plt.title("Histogram with KDE of Total Bill")
plt.xlabel("Total Bill")
plt.ylabel("Frequency")
plt.show()
```

```python
# Program 9: Create a box plot

import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

# Box plot of total bill by day
sns.boxplot(x="day", y="total_bill", data=df)

plt.title("Box Plot of Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
# Program 10: Create a violin plot

import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

# Violin plot of total bill by day
sns.violinplot(x="day", y="total_bill", data=df)

plt.title("Violin Plot of Total Bill by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")
plt.show()
```

```python
# Program 11: Create a pair plot

import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

# Pair plot for numeric columns
sns.pairplot(df)

plt.show()
```

```python
# Program 12: Create a heatmap using correlation matrix

import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

# Select only numeric columns
numeric_df = df.select_dtypes(include=["number"])

# Create correlation matrix
corr = numeric_df.corr()

# Display heatmap
sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.show()
```

```python
# Program 13: Use hue to compare categories

import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset("tips")

# Scatter plot with hue
sns.scatterplot(x="total_bill", y="tip", hue="sex", data=df)

plt.title("Total Bill vs Tip by Gender")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

```python
# Program 15: Seaborn with a custom dataset

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Create custom student dataset
data = {
    "Name": ["Amit", "Riya", "Sohan", "Mina", "Rahul"],
    "Marks": [78, 85, 90, 88, 76],
    "Study_Hours": [2, 3, 4, 3.5, 2.5]
}

df = pd.DataFrame(data)

# Scatter plot
sns.scatterplot(x="Study_Hours", y="Marks", data=df)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

Short explanations:

lineplot() → shows trend
barplot() → shows average/comparison
countplot() → counts categories
scatterplot() → shows relation between two values
histplot() → shows distribution
boxplot() → shows spread and outliers
violinplot() → shows distribution shape
pairplot() → compares many numeric columns
heatmap() → shows correlations visually
