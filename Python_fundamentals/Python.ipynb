**9.Student grade calculator**

```
# 1. Collect Student Details
student_name = input("Enter Student Name: ")

# 2. Input marks for 3 subjects
sub1 = int(input("Enter marks for Subject 1: "))
sub2 = int(input("Enter marks for Subject 2: "))
sub3 = int(input("Enter marks for Subject 3: "))

# 3. Calculations
total = sub1 + sub2 + sub3
percentage = (total / 300) * 100

# 4. Grade Calculation Logic
if percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "Fail (F)"

# 5. Display the Final Report Card
print("\n" + "★" * 30)
print(f"      REPORT CARD: {student_name.upper()} ")
print("★" * 30)
print(f"Total Marks: {total} / 300")
print(f"Percentage:  {round(percentage, 2)}%")
print(f"Final Grade: {grade}")
print("★" * 30)
```
