# Matplotlib: visualization

# Line Diagram- plot()


```python
# Module:

# matplotlib, seaborn: visualization (bar, pie, scatter, line graph)
# numpy: numerical python
# pandas: data handling

import matplotlib.pyplot as plt
import numpy as np

xpoints = np.array([0, 6])
ypoints = np.array([0, 250])

plt.plot(xpoints, ypoints, color='red')
plt.show()

# list: ["Madhu", 24, 9.5, True] # multiple data at a time
# array(homogeneous): [10, 20, 30, 40] # only one data at a time (int)
# array(homogeneous): [10.4, 20.3, 30.5] # float
# array(homogeneous): ["Madhu", "IU", "AI"] #str
```

# pie chart- pie()

```python
# pie chart
import matplotlib.pyplot as plt
import numpy as np

y = np.array([35, 25, 25, 15])

plt.pie(y)
plt.show()
```

# Histogram- hist()

```python
# histogram
import matplotlib.pyplot as plt
import numpy as np

x = np.random.normal(170, 10, 250)

plt.hist(x)
plt.show()
```

# Bar Graph - bar()

```python
# bar graph

import matplotlib.pyplot as plt
import numpy as np

x = np.array(["A", "B", "C", "D"])
y = np.array([3, 8, 1, 10])

plt.bar(x,y, color='hotpink')
plt.show()
```

# Scatter- scatter()


```python
# scatter plot


import matplotlib.pyplot as plt
import numpy as np

x = np.array([5,7,8,7,2,17,2,9,4,11,12,9,6])
y = np.array([99,86,87,88,111,86,103,87,94,78,77,85,86])

plt.scatter(x, y)
plt.show()
```

# Grid

```python
# grid

import numpy as np
import matplotlib.pyplot as plt

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.title("Sports Watch Data")
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")

plt.plot(x, y,  color='red')

plt.grid()

plt.show()
```
