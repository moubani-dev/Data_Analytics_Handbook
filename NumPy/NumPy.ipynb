```python
# NumPy introduction

# NumPy is used for fast numerical calculations
# NumPy array is similar to a Python list, but more powerful

import numpy as np

# Create a NumPy array
arr = np.array([10, 20, 30, 40, 50])

print("NumPy Array:")
print(arr)

print("\nType of arr:")
print(type(arr))
```

```python
# 1D and 2D arrays

# 1D array = one row of values
# 2D array = rows and columns like table

import numpy as np

# 1D array
arr1 = np.array([1, 2, 3, 4, 5])

# 2D array (Row then Column)
arr2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("1D Array:")
print(arr1)

print("\n2D Array:")
print(arr2)
```

```python
# Shape and dimension

# shape tells rows and columns
# ndim tells dimension

import numpy as np

arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Shape of arr1:", arr1.shape)
print("Dimension of arr1:", arr1.ndim)

print("Shape of arr2:", arr2.shape)
print("Dimension of arr2:", arr2.ndim)
```

```python
# Indexing in NumPy

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Full array:")
print(arr)

print("\nFirst element:", arr[0])
print("Third element:", arr[2])
print("Last element:", arr[-1])
```

```python
# Slicing in NumPy

import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60])

print("Original array:")
print(arr)

print("\nElements from index 1 to 4:")
print(arr[1:5])

print("\nFirst 3 elements:")
print(arr[:3])

print("\nLast 2 elements:")
print(arr[-2:])
```

```python
# 2D array indexing

# first value is row index
# second value is column index

import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("2D Array:")
print(arr)

print("\nElement at row 0, column 1:", arr[0, 1])
print("Element at row 1, column 2:", arr[1, 2])
```

```python
# Mathematical operations

# NumPy can do element-wise operations very easily

import numpy as np

arr1 = np.array([10, 20, 30])
arr2 = np.array([1, 2, 3])

print("Array 1:", arr1)
print("Array 2:", arr2)

print("\nAddition:")
print(arr1 + arr2)

print("\nSubtraction:")
print(arr1 - arr2)

print("\nMultiplication:")
print(arr1 * arr2)

print("\nDivision:")
print(arr1 / arr2)
```

```python
# Useful NumPy functions

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Array:", arr)

print("\nSum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Maximum:", np.max(arr))
print("Minimum:", np.min(arr))
```

```python
# Special arrays

import numpy as np

# Array of zeros
a = np.zeros(5)

# Array of ones
b = np.ones(5)

# Range of values
c = np.arange(1, 11)

print("Zeros array:")
print(a)

print("\nOnes array:")
print(b)

print("\nRange array:")
print(c)
```

```python
# shape, ndim, size

import numpy as np

# Create a 2D array
arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Array:")
print(arr)

# shape = number of rows and columns
print("\nShape:", arr.shape)

# ndim = number of dimensions
print("Number of dimensions:", arr.ndim)

# size = total number of elements
print("Total number of elements:", arr.size)
```

```python
# Data Types with dtype and type conversion

import numpy as np

# Create array of integers
arr = np.array([1, 2, 3, 4, 5])

print("Original array:")
print(arr)

# Check data type
print("\nData type:", arr.dtype)

# Convert integer array to float array
arr_float = arr.astype(float)

print("\nConverted to float:")
print(arr_float)
print("New data type:", arr_float.dtype)
```

```python
# zeros(), ones(), arange(), linspace()

import numpy as np

# Array of zeros
a = np.zeros(5)
print("zeros():")
print(a)

# Array of ones
b = np.ones(5)
print("\nones():")
print(b)

# Range of values
c = np.arange(1, 11)
print("\narange():")
print(c)

# Evenly spaced values
d = np.linspace(0, 10, 5)
print("\nlinspace():")
print(d)
```

```python
# reshaping with reshape()

import numpy as np

# Create 1D array
arr = np.array([1, 2, 3, 4, 5, 6])

print("Original array:")
print(arr)

# Reshape into 2 rows and 3 columns
reshaped = arr.reshape(2, 3)

print("\nReshaped array (2 x 3):")
print(reshaped)
```

```python
# flattening with flatten() / ravel()

import numpy as np

# Create 2D array
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print("Original 2D array:")
print(arr)

# Convert into 1D array using flatten()
flat1 = arr.flatten()
print("\nFlatten using flatten():")
print(flat1)

# Convert into 1D array using ravel()
flat2 = arr.ravel()
print("\nFlatten using ravel():")
print(flat2)
```

```python
# boolean masking

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Original array:")
print(arr)

# Create boolean mask
mask = arr > 25

print("\nBoolean mask (arr > 25):")
print(mask)

# Use mask to get values
print("\nValues greater than 25:")
print(arr[mask])
```

```python
# conditional filtering

import numpy as np

arr = np.array([5, 12, 18, 25, 30, 7])

print("Original array:")
print(arr)

# Filter values greater than 15
filtered = arr[arr > 15]

print("\nFiltered values greater than 15:")
print(filtered)
```

```python
# broadcasting


import numpy as np

arr = np.array([10, 20, 30, 40])

print("Original array:")
print(arr)

# Add one number to all elements
result = arr + 5

print("\nAfter adding 5 to all elements:")
print(result)

# Multiply all elements by 2
result2 = arr * 2

print("\nAfter multiplying by 2:")
print(result2)
```

```python
# axis concept

import numpy as np

arr = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Array:")
print(arr)

# Sum column-wise
print("\nSum with axis=0 (column-wise):")
print(np.sum(arr, axis=0))

# Sum row-wise
print("\nSum with axis=1 (row-wise):")
print(np.sum(arr, axis=1))
```

```python
# statistical functions: mean(), median(), std(), var()

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Array:")
print(arr)

print("\nMean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
print("Variance:", np.var(arr))
```

```python
# min(), max(), sum(), cumsum()


import numpy as np

arr = np.array([10, 20, 30, 40, 50])

print("Array:")
print(arr)

print("\nMinimum:", np.min(arr))
print("Maximum:", np.max(arr))
print("Sum:", np.sum(arr))
print("Cumulative Sum:", np.cumsum(arr))
```

```python
# sorting with sort()

import numpy as np

arr = np.array([40, 10, 30, 20, 50])

print("Original array:")
print(arr)

# Sort array
sorted_arr = np.sort(arr)

print("\nSorted array:")
print(sorted_arr)
```

```python
# copying vs viewing

import numpy as np

arr = np.array([10, 20, 30, 40])

print("Original array:")
print(arr)

# View shares memory
view_arr = arr.view()

# Copy creates separate memory
copy_arr = arr.copy()

# Change original array
arr[0] = 999

print("\nAfter changing original array:")
print("Original array:", arr)
print("View array:", view_arr)
print("Copy array:", copy_arr)
```

```python
# random module

import numpy as np

# Random integer values
rand_int = np.random.randint(1, 10, size=5)
print("Random integers:")
print(rand_int)

# Random float values between 0 and 1
rand_float = np.random.rand(5)
print("\nRandom floats:")
print(rand_float)

# Random 2D array
rand_2d = np.random.randint(1, 100, size=(2, 3))
print("\nRandom 2D array:")
print(rand_2d)
```

```python
# stacking arrays

import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Array a:")
print(a)

print("\nArray b:")
print(b)

# Horizontal stacking
h_stack = np.hstack((a, b))
print("\nHorizontal stack:")
print(h_stack)

# Vertical stacking
v_stack = np.vstack((a, b))
print("\nVertical stack:")
print(v_stack)
```

```python
# splitting arrays

import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60])

print("Original array:")
print(arr)

# Split into 3 equal parts
split_arr = np.split(arr, 3)

print("\nSplit array into 3 parts:")
print(split_arr)
```

```python
# universal functions (sqrt, log, exp, etc.)


import numpy as np

arr = np.array([1, 4, 9, 16])

print("Array:")
print(arr)

print("\nSquare root:")
print(np.sqrt(arr))

print("\nExponential:")
print(np.exp(arr))

print("\nLogarithm:")
print(np.log(arr))
```

```python
# handling missing values basics (nan)

import numpy as np

arr = np.array([10, 20, np.nan, 40, 50])

print("Array with missing value:")
print(arr)

# Check missing values
print("\nIs NaN present?")
print(np.isnan(arr))

# Mean ignoring NaN
print("\nMean ignoring NaN:")
print(np.nanmean(arr))

# Sum ignoring NaN
print("\nSum ignoring NaN:")
print(np.nansum(arr))
```

```python
# basic linear algebra functions

import numpy as np

a = np.array([
    [1, 2],
    [3, 4]
])

b = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix A:")
print(a)

print("\nMatrix B:")
print(b)

# Matrix multiplication
print("\nMatrix multiplication:")
print(np.dot(a, b))

# Transpose
print("\nTranspose of A:")
print(a.T)

# Determinant
print("\nDeterminant of A:")
print(np.linalg.det(a))
```

```python
# practical NumPy for data analytics workflow

import numpy as np

# Sales data of 7 days
sales = np.array([200, 250, 180, 300, 280, 220, 260])

print("Sales data:")
print(sales)

# Basic analysis
print("\nTotal sales:", np.sum(sales))
print("Average sales:", np.mean(sales))
print("Maximum sales:", np.max(sales))
print("Minimum sales:", np.min(sales))

# Days where sales are greater than 240
print("\nSales greater than 240:")
print(sales[sales > 240])

# Add bonus 20 to all days
updated_sales = sales + 20
print("\nUpdated sales after bonus:")
print(updated_sales)

# Sort sales
print("\nSorted sales:")
print(np.sort(sales))
```
